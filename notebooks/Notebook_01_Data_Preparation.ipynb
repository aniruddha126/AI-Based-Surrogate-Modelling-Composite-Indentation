```python
# ======================================================================
# CELL-01 : IMPORT LIBRARIES
# ======================================================================

import pandas as pd
import numpy as np

from sklearn.preprocessing import MinMaxScaler

from sklearn.model_selection import train_test_split

import joblib

print("Libraries Imported Successfully")
```

```python
# ======================================================================
# CELL-02 : CASE FILE DEFINITIONS
# ======================================================================

case_files = {

    "Case_01": "Case_01_200x200_10mm.csv",
    "Case_02": "Case_02_200x200_15mm.csv",
    "Case_03": "Case_03_200x200_20mm.csv",

    "Case_04": "Case_04_250x250_10mm.csv",
    "Case_05": "Case_05_300x300_10mm.csv",

    "Case_06": "Case_06_250x250_15mm.csv",
    "Case_07": "Case_07_250x250_20mm.csv",

    "Case_08": "Case_08_300x300_15mm.csv",
    "Case_09": "Case_09_300x300_20mm.csv",

    "Case_10": "Case_10_350x350_10mm.csv",
    "Case_11": "Case_11_350x350_15mm.csv",
    "Case_12": "Case_12_350x350_20mm.csv",

    "Case_13": "Case_13_225x225_12mm.csv",
    "Case_14": "Case_14_225x225_18mm.csv",

    "Case_15": "Case_15_275x275_12mm.csv",
    "Case_16": "Case_16_275x275_18mm.csv",

    "Case_17": "Case_17_325x325_12mm.csv",
    "Case_18": "Case_18_325x325_18mm.csv",

    "Case_19": "Case_19_300x300_5mm.csv",
    "Case_20": "Case_20_300x300_25mm.csv",

    "Case_21": "Case_21_400x400_10mm.csv",
    "Case_22": "Case_22_400x400_20mm.csv",

    "Case_23": "Case_23_250x250_5mm.csv",
    "Case_24": "Case_24_250x250_25mm.csv",

    "Case_25": "Case_25_375x375_10mm.csv",
    "Case_26": "Case_26_375x375_20mm.csv",
    "Case_27": "Case_27_400x400_15mm.csv"
}
```

```python
# ======================================================================
# CELL-03 : CASE METADATA
# ======================================================================

case_metadata = {

    "Case_01":{"PlateSize":200,"IndentationDepth":10},
    "Case_02":{"PlateSize":200,"IndentationDepth":15},
    "Case_03":{"PlateSize":200,"IndentationDepth":20},

    "Case_04":{"PlateSize":250,"IndentationDepth":10},
    "Case_05":{"PlateSize":300,"IndentationDepth":10},

    "Case_06":{"PlateSize":250,"IndentationDepth":15},
    "Case_07":{"PlateSize":250,"IndentationDepth":20},

    "Case_08":{"PlateSize":300,"IndentationDepth":15},
    "Case_09":{"PlateSize":300,"IndentationDepth":20},

    "Case_10":{"PlateSize":350,"IndentationDepth":10},
    "Case_11":{"PlateSize":350,"IndentationDepth":15},
    "Case_12":{"PlateSize":350,"IndentationDepth":20},

    "Case_13":{"PlateSize":225,"IndentationDepth":12},
    "Case_14":{"PlateSize":225,"IndentationDepth":18},

    "Case_15":{"PlateSize":275,"IndentationDepth":12},
    "Case_16":{"PlateSize":275,"IndentationDepth":18},

    "Case_17":{"PlateSize":325,"IndentationDepth":12},
    "Case_18":{"PlateSize":325,"IndentationDepth":18},

    "Case_19":{"PlateSize":300,"IndentationDepth":5},
    "Case_20":{"PlateSize":300,"IndentationDepth":25},

    "Case_21":{"PlateSize":400,"IndentationDepth":10},
    "Case_22":{"PlateSize":400,"IndentationDepth":20},

    "Case_23":{"PlateSize":250,"IndentationDepth":5},
    "Case_24":{"PlateSize":250,"IndentationDepth":25},

    "Case_25":{"PlateSize":375,"IndentationDepth":10},
    "Case_26":{"PlateSize":375,"IndentationDepth":20},
    "Case_27":{"PlateSize":400,"IndentationDepth":15}
}
```

```python
# ======================================================================
# CELL-04 : INDEPENDENT TEST CASES
# ======================================================================

test_cases = [

    "Case_20",

    "Case_22"

]
```

```python
# ======================================================================
# CELL-05 : DEVELOPMENT CASES
# ======================================================================

development_cases = [

    case

    for case in case_files.keys()

    if case not in test_cases

]

print("Development Cases :", len(development_cases))

print("Independent Test Cases :", len(test_cases))
```

```python
# ======================================================================
# CELL-06 : LOAD DEVELOPMENT DATA
# ======================================================================

development_frames = []

for case in development_cases:

    df = pd.read_csv(
        case_files[case]
    )

    df["PlateSize"] = case_metadata[case]["PlateSize"]

    df["IndentationDepth"] = case_metadata[case]["IndentationDepth"]

    df["Case"] = case

    development_frames.append(df)

development_df = pd.concat(

    development_frames,

    ignore_index=True

)

print(
    development_df.shape
)
```

```python
# ======================================================================
# CELL-07 : LOAD TEST DATA
# ======================================================================

test_frames = []

for case in test_cases:

    df = pd.read_csv(
        case_files[case]
    )

    df["PlateSize"] = case_metadata[case]["PlateSize"]

    df["IndentationDepth"] = case_metadata[case]["IndentationDepth"]

    df["Case"] = case

    test_frames.append(df)

test_df = pd.concat(

    test_frames,

    ignore_index=True

)

print(
    test_df.shape
)
```

```python
# ======================================================================
# CELL-08 : INPUTS AND OUTPUTS
# ======================================================================

INPUTS = [

    "Y",
    "Z",
    "PlateSize",
    "IndentationDepth"

]

OUTPUTS = [

    "Displacement",

    "Stress_L1",

    "Stress_L2",

    "Stress_L4"

]
```

```python
# ======================================================================
# CELL-09 : CREATE FEATURES
# ======================================================================

X = development_df[INPUTS].values

Y = development_df[OUTPUTS].values

X_test = test_df[INPUTS].values

Y_test = test_df[OUTPUTS].values

print(X.shape)

print(Y.shape)
```

```python
# ======================================================================
# CELL-10 : SCALE INPUTS
# ======================================================================

input_scaler = MinMaxScaler()

X_scaled = input_scaler.fit_transform(X)

X_test_scaled = input_scaler.transform(X_test)

print("Input Scaling Completed")
```

```python
# ======================================================================
# CELL-11 : SCALE OUTPUTS
# ======================================================================

output_scaler = MinMaxScaler()

Y_scaled = output_scaler.fit_transform(Y)

Y_test_scaled = output_scaler.transform(Y_test)

print("Output Scaling Completed")
```

```python
# ======================================================================
# CELL-12 : TRAIN VALIDATION SPLIT
# ======================================================================

X_train, X_val, Y_train, Y_val = train_test_split(

    X_scaled,

    Y_scaled,

    test_size=0.20,

    random_state=42,

    shuffle=True

)

print("Training Shape")

print(X_train.shape)

print(Y_train.shape)

print()

print("Validation Shape")

print(X_val.shape)

print(Y_val.shape)
```

```python
# ======================================================================
# CELL-13 : SAVE DATASETS
# ======================================================================

np.save("X_train.npy", X_train)
np.save("Y_train.npy", Y_train)

np.save("X_val.npy", X_val)
np.save("Y_val.npy", Y_val)

np.save("X_test.npy", X_test_scaled)
np.save("Y_test.npy", Y_test_scaled)

print("Datasets Saved")
```

```python
# ======================================================================
# CELL-14 : SAVE SCALERS
# ======================================================================

joblib.dump(

    input_scaler,

    "input_scaler.pkl"

)

joblib.dump(

    output_scaler,

    "output_scaler.pkl"

)

print("Scalers Saved")
```

```python
# ======================================================================
# CELL-15 : SAVE TEST DATASET
# ======================================================================

test_df.to_csv(

    "Independent_Test_Dataset.csv",

    index=False

)

print(
    "Independent Test Dataset Saved"
)
```

```python
# ======================================================================
# CELL-16 : SAVE METADATA
# ======================================================================

metadata = {

    "Development_Cases":
    development_cases,

    "Independent_Test_Cases":
    test_cases,

    "Random_State":
    42,

    "Validation_Split":
    0.20

}

joblib.dump(

    metadata,

    "dataset_metadata.pkl"

)

print("Metadata Saved")
```
